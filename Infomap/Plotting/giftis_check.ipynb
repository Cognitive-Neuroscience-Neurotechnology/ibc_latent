```python
import os, numpy as np, nibabel as nib

subject = "04"     # two-digit
working_dir = "/ptmp/hmueller2/Downloads"
subdir = os.path.join(working_dir, f"subnetworks/infomap/sub-{subject}")
fpn_file = os.path.join(subdir, "FPN_communities.dscalar.nii")

# Vertex area file (used in your MATLAB Step 7)
va_file = os.path.join(
    working_dir, f"fmriprep_out/sub-{subject}/anat",
    f"sub-{subject}.midthickness_va.32k_fs_LR.dscalar.nii"
)

# Load labels (0 = background, >0 = community IDs)
lab = nib.load(fpn_file).get_fdata().astype(int).ravel()

# Load vertex areas for cortex mask and area-based %s
va = nib.load(va_file).get_fdata().ravel()
cortex_mask = va > 0

total_cortex_verts = int(cortex_mask.sum())
total_cortex_area = float(va[cortex_mask].sum())

ids = [i for i in np.unique(lab) if i != 0]
print(f"Total cortex verts: {total_cortex_verts}, total cortex area: {total_cortex_area:.2f} mm^2")

for i in ids:
    m = lab == i
    nverts = int(m.sum())
    pct_verts = 100.0 * nverts / total_cortex_verts
    # Only consider the 64k cortex vertices for area calculation
    area = float(va[(lab[:64984] == i)].sum())
    pct_area = 100.0 * area / total_cortex_area
    print(f"Community {i}: {nverts} verts ({pct_verts:.3f}% cortex by count), {area:.2f} mm^2 ({pct_area:.3f}% cortex by area)")
```

```python
import os, numpy as np, nibabel as nib

subject = "15"  # two-digit
working_dir = "/ptmp/hmueller2/Downloads"
subdir = os.path.join(working_dir, f"subnetworks/infomap/sub-{subject}")
fpn_file = os.path.join(subdir, "FPN_communities.dscalar.nii")

#print("For subject ", subject)

# Optional: area-based percentages within FPN
va_file = os.path.join(
    working_dir, f"fmriprep_out/sub-{subject}/anat",
    f"sub-{subject}.midthickness_va.32k_fs_LR.dscalar.nii"
)

# Load labels (0 = non-FPN, >0 = FPN community IDs)
lab = nib.load(fpn_file).get_fdata().astype(np.int64).ravel()
ids = [i for i in np.unique(lab) if i != 0]

# Percent of FPN by vertex count
fpn_mask = lab > 0
total_fpn_verts = int(fpn_mask.sum())
print(f"Total FPN verts: {total_fpn_verts}")

for i in ids:
    nverts = int((lab == i).sum())
    pct_of_fpn = 100.0 * nverts / total_fpn_verts if total_fpn_verts > 0 else 0.0
    print(f"Community {i}: {nverts} verts ({pct_of_fpn:.3f}% of FPN verts)")

# Optional: percent of FPN by surface area
if os.path.isfile(va_file):
    va = nib.load(va_file).get_fdata().astype(np.float64).ravel()
    cortex_mask = va > 0                 # 64k cortex vertices
    # Only use the first 64984 elements of fpn_mask to match cortex_mask
    fpn_cortex = fpn_mask[:64984] & cortex_mask  # FPN restricted to cortex
    total_fpn_area = float(va[fpn_cortex].sum())
    print(f"\nTotal FPN area (cortex): {total_fpn_area:.2f} mm^2")

    for i in ids:
        m = (lab == i) & cortex_mask
        area = float(va[m].sum())
        pct_area_of_fpn = 100.0 * area / total_fpn_area if total_fpn_area > 0 else 0.0
        print(f"Community {i}: {area:.2f} mm^2 ({pct_area_of_fpn:.3f}% of FPN area)")
else:
    print("\nVertex area file not found; skipping area-based percentages.")
```
