```python
import nibabel as nib
import os
import glob

# Base directory containing session folders
base_directory = '/ptmp/hmueller2/Downloads/fmriprep_out/sub-01/'

# Iterate through all session folders
for session_folder in glob.glob(os.path.join(base_directory, 'ses-*', 'postfmriprep', 'GLM')):
    print(f"Checking files in: {session_folder}")
    for filename in os.listdir(session_folder):
        filepath = os.path.join(session_folder, filename)
        if filepath.endswith('.nii') or filepath.endswith('.nii.gz'):  # Check for NIfTI files
            img = nib.load(filepath)
            if img.get_fdata().shape[1:] != (91282,):
                print(f"File {filename} in {session_folder} does not have the expected shape.")
```

```python
import os
import nibabel as nib
import glob

Subject = '11'
TR = 2  # Repetition time in seconds

# Base directory for the subject
base_path = f'/ptmp/hmueller2/Downloads/fmriprep_out/sub-{Subject}/'

# Find all .dtseries.nii files in /ses-*/postfmriprep/GLM/ directories
files = glob.glob(os.path.join(base_path, 'ses-*', 'postfmriprep', 'GLM', '*.dtseries.nii'))

# Check the shape of the files and add the first number (timepoints) together
n_timepoints = sum(nib.load(file).get_fdata().shape[0] for file in files)
print('Total number of timepoints:', n_timepoints)

# Calculate total minutes
total_minutes = (n_timepoints * TR) / 60
print('Total minutes: {:.2f}'.format(total_minutes))
```
