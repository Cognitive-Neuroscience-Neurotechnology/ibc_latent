```python
import nibabel as nib
import matplotlib.pyplot as plt
import numpy as np
```

### Raw bold

```python
from nilearn.plotting import view_img

# Path to your NIfTI file
nii_path = "/home/hmueller2/ptmp/Downloads/ibc_raw_test/sub-01/ses-00/func/sub-01_ses-00_task-ArchiStandard_dir-pa_bold.nii.gz"

# Load the NIfTI image
img = nib.load(nii_path)
data = img.get_fdata()

# Plot the middle slice of the first volume (if 4D)
if data.ndim == 4:
    slice_data = data[:, :, data.shape[2] // 2, 0]
else:
    slice_data = data[:, :, data.shape[2] // 2]

plt.imshow(np.rot90(slice_data), cmap="gray")
plt.title("Middle Slice")
plt.axis("off")
plt.show()
```
